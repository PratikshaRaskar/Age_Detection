```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# ✅ Load model (important fix)
model = load_model('models/age_detect_cnn_model.h5', compile=False)

# ✅ Load face detector
detector = cv2.CascadeClassifier('models/haarcascade_frontalface_default.xml')

# ✅ Age ranges
ranges = ['1-2','3-9','10-20','21-27','28-45','46-65','66-116']

# ✅ Start webcam
cam = cv2.VideoCapture(1)

while True:
    ret, frame = cam.read()
    
    if not ret:
        print("Failed to grab frame")
        break

    # Convert to grayscale for detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = detector.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        face = gray[y:y+h, x:x+w]

        # Resize to model input size
        face = cv2.resize(face, (200, 200))

        # Normalize
        face = face / 255.0

        # Reshape for model
        face = np.reshape(face, (1, 200, 200, 1))

        # Predict age
        age_pred = model.predict(face, verbose=0)
        age_range = ranges[np.argmax(age_pred)]

        # Draw rectangle
        cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 2)

        # Label background
        cv2.rectangle(frame, (x, y+h), (x+w, y+h+40), (255, 0, 0), -1)

        # Put text
        cv2.putText(frame, age_range, (x+10, y+h+30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,255,255), 2)

    # Show output
    cv2.imshow('Age Detection', frame)

    # Press ESC to exit
    if cv2.waitKey(1) == 27:
        break

# Release resources
cam.release()
cv2.destroyAllWindows()
```
